# Live code from session

```python
# for x, generate a sequence from 0 to 49
# 
np.random.seed(613)
x = np.arange(50) # generate a sequence from 0 to 49
y = np.random.randint(0, 100,50) # generate 50 random numbers between 0 and 99
```

```python
fig, ax = plt.subplots(figsize=(5, 3))

# plot data
ax.scatter(x,y, marker="*", color = "indigo")
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y, marker="*", color = "#7425b9", linestyle="--", linewidth=2, markeredgecolor = '#fa9359', markerfacecolor = '#fa9359')
ax.grid(axis = 'y', color='green', linewidth=3, linestyle='-.')
```

```python
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

# plot data
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)

# add labels
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```
